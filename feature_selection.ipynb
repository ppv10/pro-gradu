# Exploratory feature associations

This notebook describes the supplied cohort. Its full-data associations are exploratory and are **not** used to select predictors or estimate predictive performance. Feature selection for classification occurs within inner cross-validation folds in the model workflow.

The binary target records whether a death event was observed during the recorded follow-up. Follow-up duration (`time`) is excluded from baseline classification features.

```python
from pathlib import Path
import pandas as pd

from pro_gradu.data import BASELINE_FEATURES, load_dataset
from pro_gradu.classification import classification_features

data_path = Path('heart_failure_clinical_records_dataset.csv')
df = load_dataset(data_path)
X, y = classification_features(df)
print(f'{len(df)} records; {int(y.sum())} observed events ({y.mean():.1%})')
df.head()
```

## Cohort summaries

The table gives basic summaries for the baseline predictors. Some predictors are binary indicators. These summaries do not account for confounding, follow-up duration, or censoring.

```python
X.describe().T
```

## Rank associations with observed events

Spearman correlations summarize marginal monotonic associations with the binary observed-event indicator. They do not establish causality and are not a feature-selection result. Use the nested cross-validation workflow for predictive evaluation.

```python
associations = X.assign(DEATH_EVENT=y).corr(method='spearman')['DEATH_EVENT'].drop('DEATH_EVENT')
associations.reindex(associations.abs().sort_values(ascending=False).index).rename('spearman_rho').to_frame()
```

## Limitations

This is a small observational cohort (299 records). Marginal associations can be unstable and can differ from adjusted model effects. This notebook does not tune models, choose features for a final model, or make clinical recommendations. The source data does not provide patient identifiers; identical rows are not removed as presumed duplicates.
