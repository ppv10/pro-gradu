It now checks the cohort and records the classifier families used in the nested comparison. Run `hyperparameter_tuning_and_final_models.ipynb` for the actual evaluation. No predictive score is reported here.

```python
from pathlib import Path
import pandas as pd

from pro_gradu.data import load_dataset
from pro_gradu.evaluation import default_model_specs

df = load_dataset(Path('heart_failure_clinical_records_dataset.csv'))
print(f'Records: {len(df)}')
print(f'Observed events: {int(df.DEATH_EVENT.sum())} ({df.DEATH_EVENT.mean():.1%})')
print(f'No observed event by end of follow-up: {int((df.DEATH_EVENT == 0).sum())}')
df.info()
```

These are descriptive checks only. `DEATH_EVENT = 0` means no death event was observed during recorded follow-up; the label does not imply that the patient never died.

## Candidate classifiers

The nested benchmark evaluates each model family with and without mutual-information feature selection. Both variants use the same outer folds. Standardization, selection, and hyperparameter tuning are fitted only within training folds.

```python
specs = default_model_specs()
roster = pd.DataFrame([{'model': s.name, 'feature_selection': s.select_features, 'search_iterations': s.n_iter} for s in specs])
roster.drop_duplicates('model').reset_index(drop=True)
```

## Interpretation

The dataset has 299 records and is observational. This analysis estimates prediction of an event observed under the source cohort's follow-up process, not lifetime mortality or causal effects. The broad model roster is an exploratory comparison and does not establish a clinically optimal algorithm.
